## 5.2 Evaluate model performance with metrics

Metrics used 
- **MAE (Mean Absolute Error)** : average absolute difference between predictions and true values
  - Lower MAE = better predictions
- **RMSE (Root Mean Squared Error)** : average difference between predicted values and actual values.
  - Lower RMSE = better predictions, smaller errors
- **R2 (R-squared)** : proportion of the variance explained by the model
  - Value close to 1 = better fit 

Results were rounded in this document MAE and RMSE: no decimals. R^2 with three decimals.

### 5.2.1 Error metrics for 'price' prediction

Results for predicting 'price' 

Ridge – 'price' 

| MAE | RMSE | R² |
|-----|------|------|
| 4731 | 6813 | 0.787 |
 
LightGBM – 'price' 

| MAE | RMSE | R² |
|-----|------|------|
| 1943 | 2947 | 0.960 |

**Conclusions**
- Ridge 'price' predictions are off by about 4700 euros whereas LightGBM is only off by almost 2000 euros
- LightGBM makes fewer large errors as RMSE shows
- LightGBM explains 96% of the variance in prices which makes it a stronger fit than Ridge with only 79%
- The price depends on non-linear relationships between different features which LightGM is able to capture much better than Ridge linear regression model
- Price predictions could be used
