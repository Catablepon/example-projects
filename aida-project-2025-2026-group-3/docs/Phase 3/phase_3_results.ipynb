```python
cars_df3['make_name'].nunique() # number of unique values in the column
```

```python
# cut unique values to list top2 makers, replace rest with the value 'other'

top_20 = top_20[:20].index # list of top20

def top_20_make(make): # function to return top 20
    if make in top_20:
        return make
    else:
        return 'Other'

cars_df3['make_name'] = cars_df3['make_name'].apply(top_20_make) # apply function
cars_df3['make_name'].value_counts() # result
```

### One-hot encoding below

```python
# Lista pienistä kategorisista sarakkeista (alle ~20 uniikkia, parissa oli 21)
cat_cols = [
    'body_type', 'fuel_type', 'transmission', 'wheel_system', 
    'car_manufacturer', 'model_name', 'interior_color', 
    'exterior_color', 'engine_type'
]

# Looppi, joka muuntaa jokaisen sarakkeen dummyiksi
for col in cat_cols:
    dummies = pd.get_dummies(cars_df4[col], prefix=col, drop_first=True)
    cars_df4 = pd.concat([cars_df4, dummies], axis=1)
    cars_df4.drop(col, axis=1, inplace=True)
```

```python
dummy_cols = cars_df4.select_dtypes(include=['bool']).columns
cars_df4[dummy_cols] = cars_df4[dummy_cols].astype(int)
```
